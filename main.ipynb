## Heart Failure Data Visualization 

Heart failure occurs when the heart is unable to pump blood effectively, leading to serious health complications. Understanding the factors that contribute to heart failure can help in identifying high-risk patients early.

This project uses the Heart Failure Clinical Records Dataset from Kaggle, which contains various medical attributes of patients, such as age, blood pressure, ejection fraction, and serum creatinine levels.

The aim of this project is to analyze and visualize the dataset using Pandas, Matplotlib, and Seaborn to uncover meaningful insights and patterns related to heart failure.

The analysis includes:

Data cleaning and preprocessing

Exploring distributions of key medical features

Visualizing relationships between different clinical factors

Identifying trends that may be associated with patient survival

```python
#import libraries
import pandas as pd

#loading csv

df = pd.read_csv('dataset/heart_failure_clinical_records_dataset.csv')
```

## Distribution of patient ages by death event and sex 

 Each point represents a patient’s age, grouped by survival status and sex. The spread of points helps visualize how age varies within each group 

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.stripplot(data=df.replace({'sex':{0:'Female',1:'Male'}}),
               x='DEATH_EVENT',
               y='age',
               hue='sex',
               dodge=True, 
               alpha=0.6)


plt.xlabel("Death Event (0 = Survived, 1 = Died)")
plt.ylabel("Average Age")
plt.title("Average Age by Death Event and Sex")
plt.grid(linestyle=':',color="gray")
```

## Distribution of follow-up time (in days) for all patients in the dataset 
Most patients were followed for a shorter duration, while a few had longer observation periods. The distribution helps understand the overall study timeline 

```python
sns.histplot(df['time'], bins=20, kde=True,color='blue')
plt.xlabel('Number of follow-up days')
plt.ylabel('Number of Patients')
plt.grid(linestyle=":")
plt.title('Distribution of Follow-Up Time (Days) for Patients')
plt.show()

```

## Follow-up Period vs. Death Event
This histogram shows the distribution of patients based on their follow-up period (in days) and survival status. The **x-axis** represents the number of follow up days a patient had and the **y-axis** represents the number of patients.

```python
sns.histplot(data=df.replace({'DEATH_EVENT' : {0:'Survived',1:'Died'}}),
            x='time',
            hue='DEATH_EVENT',
            kde=True,multiple='stack',
            palette='Set1')
plt.xlabel('Follow up period(days)')
plt.ylabel('Number of patients')
plt.grid(linestyle=':')
plt.show()
```

### Ejection Fraction vs. Death Event

This boxplot compares the distribution of ejection fraction values between patients who survived and those who died.  
The **x-axis** represents the ejection fraction (%), which indicates how effectively the heart pumps blood.  
Patients who **died** generally show lower ejection fraction values compared to those who **survived**,  
suggesting that reduced cardiac function is a key factor associated with higher mortality.


```python
sns.boxplot(data=df.replace({'DEATH_EVENT': {0:'Survived',1:'Died'}}),
            x='ejection_fraction',
            hue='DEATH_EVENT',
            palette='Set2')
plt.xlabel('Ejection_fraction(%)')
plt.title('Ejection Fraction Distribution by Survival Status')
plt.grid(linestyle=':', alpha=0.6)
plt.show()
```

### Relationship between Ejection Fraction and Age

This scatter plot shows the relationship between **ejection fraction** and **age** of the patients, 
colored by their survival status.  
The **x-axis** represents ejection fraction (%), and the **y-axis** represents the patient’s age.  

From the plot, it can be observed that patients who **died** tend to have a **lower ejection fraction** 
and are generally **older**, suggesting that both reduced heart efficiency and higher age 
are important factors associated with death in heart failure patients.


```python
sns.scatterplot(data=df.replace({'DEATH_EVENT': {0:'Survived',1:'Died'}}),
            x='ejection_fraction',
            y='age',
            hue='DEATH_EVENT',
            palette=('green','orange'))
sns.set_style('whitegrid')
plt.xlabel('Ejection Fraction')
plt.ylabel('Age')
plt.title('Ejection Fraction vs Age')
plt.show()
```

### Correlation Between Lifestyle, Health Factors, and Death Event

This heatmap displays the correlation between key health and lifestyle factors — 
**diabetes**, **anaemia**, **smoking**, **sex**, and the **death event** outcome.  

Each cell shows the **correlation coefficient** (ranging from -1 to +1):  
- **Positive values** → both factors increase together.  
- **Negative values** → one factor increases as the other decreases.  
- **Values near 0** → little or no linear relationship.

From the plot, we can observe which factors have stronger associations with mortality.  
For example, a higher positive correlation with `DEATH_EVENT` may indicate that the condition 
is more common among patients who died.


```python
columns = ['diabetes', 'anaemia', 'smoking', 'sex', 'DEATH_EVENT']
df[columns].corr()
```

```python
columns = ['diabetes', 'anaemia', 'smoking', 'sex', 'DEATH_EVENT']
corr = df[columns].corr()
corr = corr.rename(index={
    'diabetes': 'Diabetes',
    'anaemia': 'Anaemia',
    'smoking': 'Smoking',
    'sex': 'Sex',
    'DEATH_EVENT': 'Death Event'
}, columns={
    'diabetes': 'Diabetes',
    'anaemia': 'Anaemia',
    'smoking': 'Smoking',
    'sex': 'Sex',
    'DEATH_EVENT': 'Death Event'
})

sns.heatmap(data=corr, annot=True, cmap='Reds')
plt.title('Heatmap showing correlation between lifestyle, health factors and death event')
plt.show()
```

### Correlation Between Clinical Measurements and Death Event

This heatmap displays the correlation between major clinical laboratory measurements — 
**creatinine phosphokinase**, **serum creatinine**, **platelets**, and **serum sodium** — 
with the **death event** outcome.

Each value represents the **correlation coefficient** between two variables:
- **Positive correlation (+1)** → both increase together  
- **Negative correlation (–1)** → one increases while the other decreases  
- **Values near 0** → weak or no linear relationship

From the heatmap, we can identify which biochemical factors show stronger relationships with mortality.  
Typically, a **higher serum creatinine** (indicating poor kidney function) or **low serum sodium** 
tends to correlate with an increased risk of death in heart failure patients.


```python
columns2 = ['creatinine_phosphokinase','serum_creatinine','platelets','serum_sodium','DEATH_EVENT']
df[columns2].corr()
```

```python
columns2 = ['creatinine_phosphokinase','serum_creatinine','platelets','serum_sodium','DEATH_EVENT']
corr2 = df[columns2].corr()
corr2 = corr2.rename(index={
    'creatinine_phosphokinase': 'Creatine Phosphokinase',
    'serum_creatinine': 'Serum Creatine',
    'platelets': 'Platelets',
    'serum_sodium': 'Serum Sodium',
    'DEATH_EVENT': 'Death Event'
}, columns={
    'creatinine_phosphokinase': 'Creatine Phosphokinase',
    'serum_creatinine': 'Serum Creatine',
    'platelets': 'Platelets',
    'serum_sodium': 'Serum Sodium',
    'DEATH_EVENT': 'Death Event'  
})

sns.heatmap(data=corr2, annot=True, cmap='Blues')
plt.title('Heatmap showing correlation between clinical measurements and death event')
plt.show()
```

### Ejection Fraction vs Serum Creatinine

The following strip plot shows the relationship between **Ejection Fraction** (x-axis) and **Serum Creatinine** (y-axis) for patients, with points colored by survival outcome (`Survived` or `Died`):

- **X-axis:** Ejection Fraction – a measure of the heart’s pumping efficiency.  
- **Y-axis:** Serum Creatinine – an indicator of kidney function.  

The plot allows visual comparison of patients’ creatinine levels across different ejection fractions and highlights trends related to survival outcomes.


```python
sns.stripplot(data=df.replace({'DEATH_EVENT': {0:'Survived',1:'Died'}})
              ,x='ejection_fraction',y='serum_creatinine',
              hue='DEATH_EVENT',
              palette='dark',
              jitter=True,dodge=True)
sns.set_style('white')
plt.xlabel('Ejection Fraction')
plt.ylabel('Serum Creatinine')
plt.title('Ejection Fraction vs Serum Creatinine')
plt.grid('whitegrid')
plt.show()
```

### Platelet Count Distribution by Survival Status

The violin plot shows the distribution of **platelet counts** for patients who **survived** versus those who **died**:

- **Y-axis:** Platelet count (numeric).  
- **Hue:** DEATH_EVENT (Survived or Died).  
- **Violin shape:** Shows the full distribution of platelet counts, including density and spread.  

This plot helps identify differences in platelet levels between survivors and deceased patients, which may indicate clinical relevance in heart failure outcomes.


```python
sns.violinplot(data=df.replace({'DEATH_EVENT' : {0:"Survived",1:"Died"}}),
               y='platelets',
               hue='DEATH_EVENT',
               palette='Set2',
               )
plt.xlabel('Platelets')
plt.grid()
plt.title('Platelet Count Distribution by Survival Status')
plt.show()
```

### Ejection Fraction vs Serum Creatinine by Blood Pressure Status

The FacetGrid scatter plot shows the relationship between **Ejection Fraction** (x-axis) and **Serum Creatinine** (y-axis) for patients, separated by **Blood Pressure Status** (`Normal` vs `High`) and colored by survival outcome (`Survived` or `Died`).
 
This plot allows us to compare how kidney function (Serum Creatinine) and heart function (Ejection Fraction) relate to survival outcomes, separately for patients with normal vs high blood pressure. It helps visualize patterns and identify potential risk factors for mortality.


```python
import seaborn as sns
import matplotlib.pyplot as plt

df_plot = df.replace({
    'DEATH_EVENT': {0: 'Survived', 1: 'Died'},
    'high_blood_pressure': {0: 'Normal BP', 1: 'High BP'}
})

g = sns.FacetGrid(df_plot, col='high_blood_pressure', hue='DEATH_EVENT', height=5)
g.map(sns.scatterplot, 'ejection_fraction', 'serum_creatinine', alpha=0.7,s=40)
g.add_legend()
g.set_titles(col_template="{col_name}") 
g.set_axis_labels("Ejection Fraction", "Serum Creatinine")

plt.show()
```

### Ejection Fraction vs Creatine Phosphokinase by Blood Pressure Status

This FacetGrid scatter plot shows the relationship between **Ejection Fraction** (x-axis) and **Creatine Phosphokinase** (y-axis) for patients, separated by **Blood Pressure Status** (`Normal BP` vs `High BP`) and colored by survival outcome (`Survived` or `Died`).

The plot helps to explore how heart function (Ejection Fraction) and muscle/heart enzyme levels (Creatine Phosphokinase) relate to patient survival, and whether these patterns differ for patients with normal versus high blood pressure.


```python
import seaborn as sns
import matplotlib.pyplot as plt

df_plot = df.replace({
    'DEATH_EVENT': {0: 'Survived', 1: 'Died'},
    'high_blood_pressure': {0: 'Normal BP', 1: 'High BP'}
})

g = sns.FacetGrid(df_plot, col='high_blood_pressure', hue='DEATH_EVENT', height=5,palette='dark')
g.map(sns.scatterplot, 'ejection_fraction', 'creatinine_phosphokinase', alpha=0.7,s=40)
g.add_legend()
g.set_titles(col_template="{col_name}") 
g.set_axis_labels("Ejection Fraction", "Creatinine Phosphokinase")
plt.show()
```

### Distribution of Serum Sodium by Survival Status

The KDE plot shows the distribution of **Serum Sodium** levels for patients, separated by survival outcome.

This plot allows us to compare the distribution of sodium levels between survivors and deceased patients. Peaks or shifts in the curves can indicate whether lower or higher sodium levels are more common in one group, potentially highlighting a clinical risk factor.


```python

sns.kdeplot(data=df_plot,
             x='serum_sodium',
            hue='DEATH_EVENT',
            fill=True,
            palette=['pink','skyblue'],
             alpha=0.5)
plt.xlabel('Serum Sodium - blood sodium levels')
plt.title('Distribution of Serum Sodium by Survival Status')
plt.show()
```

### Pairplot of Numeric Variables by Survival Status

This pairplot visualizes the relationships between key numeric variables in the dataset, separated by **survival outcome** (`Survived` or `Died`):

- **Variables included:**
  - **CPK:** Creatinine Phosphokinase
  - **Ejection Fraction (%):** Heart pumping efficiency
  - **Platelets (k/uL):** Blood platelet count
  - **Serum Creatinine (mg/dL):** Kidney function marker
  - **Serum Sodium (mEq/L):** Sodium levels in blood

This plot helps identify correlations between variables, differences in distributions between survivors and deceased patients, and potential clinical patterns or risk factors associated with heart failure outcomes.


```python

df_plot_renamed = df_plot.rename(columns={
    'creatinine_phosphokinase': 'CPK',
    'ejection_fraction': 'Ejection Fraction (%)',
    'platelets': 'Platelets (k/uL)',
    'serum_creatinine': 'Serum Creatinine (mg/dL)',
    'serum_sodium': 'Serum Sodium (mEq/L)'
})

numeric_cols = ['CPK', 'Ejection Fraction (%)', 'Platelets (k/uL)',
                'Serum Creatinine (mg/dL)', 'Serum Sodium (mEq/L)']

pairplot_fig = sns.pairplot(df_plot_renamed[numeric_cols + ['DEATH_EVENT']],
                            hue='DEATH_EVENT',
                            palette='Set2',
                            diag_kind='kde',
                            plot_kws={'alpha':0.7, 's':40})


pairplot_fig.fig.suptitle('Pairplot of Numeric Variables by Survival Status', y=1.02, fontsize=16)
plt.show()
```
